```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv(
    "../data/processed/factor_data.csv"
)
```

```python
factors = [
    "PE",
    "EPS",
    "ROE",
    "IPS",
    "MTTR",
    "Beta",
    "M_vol",
    "M_liq",
    "M_high",
    "M_skew"
]

data[factors].hist(
    figsize=(14, 10),
    bins=20
)

plt.tight_layout()
plt.show()
```

```python
corr = data[
    factors + ["M_ExRet"]
].corr()

plt.figure(
    figsize=(10,8)
)

plt.imshow(
    corr,
    aspect="auto"
)

plt.xticks(
    range(len(corr)),
    corr.columns,
    rotation=90
)

plt.yticks(
    range(len(corr)),
    corr.columns
)

plt.colorbar()

plt.title(
    "Factor Correlation Matrix"
)

plt.tight_layout()

plt.show()
```
