```python
from pathlib import Path
import os
import sys

# If notebook is being run from project/notebooks,
# move working directory back to project/
if Path.cwd().name == "notebooks":
    os.chdir("..")

ROOT = Path.cwd()

# Add project root to Python import path
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

print("Working directory:", ROOT)

import pandas as pd

from src.modeling import (
    FEATURES,
    time_split,
    make_model,
    evaluate_model
)

data = pd.read_csv(
    "data/processed/factor_data.csv"
)

train, test = time_split(
    data,
    0.7
)

X_train = train[FEATURES]

y_train = train["M_ExRet"]

print("Train shape:", train.shape)
print("Test shape:", test.shape)
print("Features:", FEATURES)
```

## OLS

```python
ols = make_model(
    "ols"
)

ols.fit(
    X_train,
    y_train
)

ols_result = evaluate_model(
    ols,
    test
)

ols_result
```

## LASSO

```python
lasso = make_model(
    "lasso"
)

lasso.fit(
    X_train,
    y_train
)

lasso_result = evaluate_model(
    lasso,
    test
)
```

## Ridge

```python
ridge = make_model(
    "ridge"
)

ridge.fit(
    X_train,
    y_train
)

ridge_result = evaluate_model(
    ridge,
    test
)
```

## ElasticNet

```python
elastic = make_model(
    "elastic"
)

elastic.fit(
    X_train,
    y_train
)

elastic_result = evaluate_model(
    elastic,
    test
)
```
